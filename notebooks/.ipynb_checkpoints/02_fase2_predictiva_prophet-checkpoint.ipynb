```python
import pandas as pd
import logging
from typing import Optional

# [EN] Logger Configuration / [ES] Configuración de logging
logging.basicConfig(
    level=logging.INFO, 
    format='%(asctime)s - %(levelname)s - %(message)s'
)

def prepare_prophet_data(file_path: str) -> Optional[pd.DataFrame]:
    """
    [EN] Loads clean data, aggregates daily sales, and formats it for Prophet (ds, y).
    [ES] Carga los datos limpios, agrega ventas diarias y los formatea para Prophet (ds, y).
    
    Args:
        file_path (str): [EN] Path to clean CSV / [ES] Ruta al CSV limpio.
        
    Returns:
        pd.DataFrame: [EN] DataFrame ready for Prophet / [ES] DataFrame listo para Prophet.
    """
    logging.info("[EN] Aggregating daily sales... / [ES] Agregando ventas diarias...")
    
    try:
        # [EN] Load the processed dataset / [ES] Cargar el dataset procesado
        df = pd.read_csv(file_path)
        
        # [EN] Ensure date column is datetime / [ES] Asegurar que la columna fecha sea datetime
        df['date'] = pd.to_datetime(df['date'])
        
        # [EN] Aggregate total sales per day / [ES] Agregar ventas totales por día
        daily_sales = df.groupby('date')['sales'].sum().reset_index()
        
        # [EN] Prophet requires strictly 'ds' (datestamp) and 'y' (value)
        # [ES] Prophet requiere estrictamente 'ds' (fecha) e 'y' (valor)
        prophet_df = daily_sales.rename(columns={'date': 'ds', 'sales': 'y'})
        
        # [EN] Sort by date ascending / [ES] Ordenar por fecha de forma ascendente
        prophet_df = prophet_df.sort_values(by='ds')
        
        logging.info(f"[EN] Data prepared successfully. Total days: {len(prophet_df)} / [ES] Datos preparados con éxito. Días totales: {len(prophet_df)}")
        return prophet_df
        
    except FileNotFoundError:
        logging.error(f"[EN] File not found. Check path: {file_path} / [ES] Archivo no encontrado. Revisa la ruta: {file_path}")
        return None
    except Exception as e:
        logging.error(f"[EN] Execution error / [ES] Error de ejecución: {str(e)}")
        return None

# --- [EN] Execution Block / [ES] Bloque de Ejecución ---
if __name__ == "__main__":
    # [EN] Path to the file we cleaned in the previous notebook
    # [ES] Ruta al archivo que limpiamos en el notebook anterior
    PROCESSED_DATA_PATH = "../data/processed/supermarket_analysis_clean.csv"
    
    df_prophet = prepare_prophet_data(PROCESSED_DATA_PATH)
    
    if df_prophet is not None:
        display(df_prophet.head())
```

```python
from prophet import Prophet
import logging
import pandas as pd

def train_and_forecast_sales(df: pd.DataFrame, days_to_predict: int = 30):
    """
    [EN] Trains Prophet model and generates future sales forecast.
    [ES] Entrena el modelo Prophet y genera el pronóstico de ventas futuras.
    
    Args:
        df (pd.DataFrame): [EN] Historic data (ds, y) / [ES] Datos históricos (ds, y).
        days_to_predict (int): [EN] Days to forecast / [ES] Días a pronosticar.
        
    Returns:
        tuple: (Prophet model, Forecast DataFrame) / (Modelo Prophet, DataFrame de pronóstico).
    """
    logging.info("[EN] Initializing Prophet model... / [ES] Inicializando modelo Prophet...")
    
    try:
        # [EN] Initialize model with 95% confidence interval
        # [ES] Inicializar modelo con 95% de intervalo de confianza (0.95)
        # [EN] Turn off daily seasonality as we only have daily totals, not hourly
        # [ES] Apagamos la estacionalidad diaria porque tenemos totales por día, no por hora
        model = Prophet(interval_width=0.95, daily_seasonality=False)
        
        # [EN] Fit the model to our historical data
        # [ES] Entrenar el modelo con nuestros datos históricos
        logging.info("[EN] Fitting model to historical data... / [ES] Entrenando el modelo...")
        model.fit(df)
        
        # [EN] Create future dates (dataframe with historic + future dates)
        # [ES] Crear fechas futuras (dataframe con fechas históricas + futuras)
        future_dates = model.make_future_dataframe(periods=days_to_predict)
        
        # [EN] Predict
        # [ES] Predecir
        logging.info(f"[EN] Forecasting next {days_to_predict} days... / [ES] Pronosticando siguientes {days_to_predict} días...")
        forecast = model.predict(future_dates)
        
        logging.info("[EN] Forecast completed successfully! / [ES] ¡Pronóstico completado con éxito!")
        return model, forecast
        
    except Exception as e:
        logging.error(f"[EN] Forecasting error / [ES] Error en pronóstico: {str(e)}")
        return None, None

# --- [EN] Execution Block / [ES] Bloque de Ejecución ---
if 'df_prophet' in locals() and isinstance(df_prophet, pd.DataFrame):
    # [EN] Predict the next 30 days of sales / [ES] Predecir los próximos 30 días de ventas
    model, forecast_df = train_and_forecast_sales(df_prophet, days_to_predict=30)
    
    if forecast_df is not None:
        # [EN] Display only the most relevant columns of the FUTURE predictions
        # [ES] Mostrar solo las columnas más relevantes de las predicciones FUTURAS
        columns_to_show = ['ds', 'yhat', 'yhat_lower', 'yhat_upper']
        display(forecast_df[columns_to_show].tail(5))
else:
    logging.warning("[EN] Missing historic data. / [ES] Faltan datos históricos.")
```

```python
import plotly.graph_objects as go
import plotly.io as pio
import logging
import pandas as pd

# [EN] Ensure iframe rendering for Jupyter / [ES] Asegurar renderizado iframe para Jupyter
pio.renderers.default = 'iframe' 

def plot_sales_forecast(forecast_data: pd.DataFrame, historic_data: pd.DataFrame) -> go.Figure:
    """
    [EN] Generates an interactive line chart combining historic sales and future predictions.
    [ES] Genera un gráfico de líneas interactivo combinando ventas históricas y predicciones futuras.
    
    Args:
        forecast_data (pd.DataFrame): [EN] Output from Prophet / [ES] Salida de Prophet.
        historic_data (pd.DataFrame): [EN] Historic actuals / [ES] Datos reales históricos.
        
    Returns:
        go.Figure: [EN] Plotly interactive figure / [ES] Figura interactiva de Plotly.
    """
    logging.info("[EN] Rendering forecast visualization... / [ES] Renderizando visualización del pronóstico...")
    
    try:
        fig = go.Figure()

        # 1. [EN] Historic Data Line (Black) / [ES] Línea de Datos Históricos (Negro)
        fig.add_trace(go.Scatter(
            x=historic_data['ds'], 
            y=historic_data['y'],
            mode='markers+lines',
            name='[EN] Actual Sales / [ES] Ventas Reales',
            line=dict(color='black', width=2),
            marker=dict(size=4)
        ))

        # 2. [EN] Future Prediction Line (Blue) / [ES] Línea de Predicción Futura (Azul)
        fig.add_trace(go.Scatter(
            x=forecast_data['ds'], 
            y=forecast_data['yhat'],
            mode='lines',
            name='[EN] Forecast (yhat) / [ES] Pronóstico (yhat)',
            line=dict(color='blue', width=3, dash='solid')
        ))

        # 3. [EN] Upper Confidence Band / [ES] Banda Superior de Confianza
        fig.add_trace(go.Scatter(
            x=forecast_data['ds'], 
            y=forecast_data['yhat_upper'],
            mode='lines',
            line=dict(width=0),
            showlegend=False,
            name='[EN] Upper Bound / [ES] Límite Superior'
        ))

        # 4. [EN] Lower Confidence Band (Creates the shaded area) / [ES] Banda Inferior de Confianza (Crea el área sombreada)
        fig.add_trace(go.Scatter(
            x=forecast_data['ds'], 
            y=forecast_data['yhat_lower'],
            mode='lines',
            line=dict(width=0),
            fillcolor='rgba(0, 0, 255, 0.2)', # [EN] Transparent blue / [ES] Azul transparente
            fill='tonexty', # [EN] Fill to the upper bound / [ES] Rellenar hasta el límite superior
            name='[EN] 95% Confidence Interval / [ES] Intervalo de Confianza del 95%'
        ))

        # [EN] Layout optimization / [ES] Optimización del diseño
        fig.update_layout(
            title='[EN] 30-Day Sales Forecast / [ES] Pronóstico de Ventas a 30 Días',
            title_font_size=20,
            title_x=0.5,
            xaxis_title='[EN] Date / [ES] Fecha',
            yaxis_title='[EN] Total Sales ($) / [ES] Ventas Totales ($)',
            hovermode='x unified', # [EN] Shows all values on hover / [ES] Muestra todos los valores al pasar el ratón
            template="plotly_white",
            legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01)
        )
        
        return fig
    
    except Exception as e:
        logging.error(f"[EN] Visualization error / [ES] Error de visualización: {str(e)}")
        raise

# --- [EN] Execution Block / [ES] Bloque de Ejecución ---
if 'forecast_df' in locals() and 'df_prophet' in locals():
    try:
        fig_forecast = plot_sales_forecast(forecast_data=forecast_df, historic_data=df_prophet)
        fig_forecast.show()
    except Exception as exec_error:
        logging.error(f"[EN] Execution failed / [ES] Fallo en la ejecución: {str(exec_error)}")
else:
    logging.warning("[EN] Required DataFrames missing. Execute previous cells. / [ES] Faltan DataFrames. Ejecuta las celdas anteriores.")
```
